# 3. Número de reseñas y calificación promedio de cada libro

```python
query = """
SELECT
    b.book_id,
    b.title,
    COALESCE(rv.review_count,0) AS review_count,
    ROUND(rt.avg_rating,2) AS average_rating
FROM books AS b

LEFT JOIN
(
    SELECT
        book_id,
        COUNT(*) AS review_count
    FROM reviews
    GROUP BY book_id
) rv
ON b.book_id = rv.book_id

LEFT JOIN
(
    SELECT
        book_id,
        AVG(rating) AS avg_rating
    FROM ratings
    GROUP BY book_id
) rt
ON b.book_id = rt.book_id

ORDER BY review_count DESC;
"""

pd.io.sql.read_sql(query, con = engine)
```

El análisis muestra que existe una gran variabilidad en la interacción de los usuarios con los libros. Algunos títulos, como Twilight (Twilight #1), The Curious Incident of the Dog in the Night-Time, Harry Potter and the Chamber of Secrets, The Da Vinci Code y The Glass Castle, destacan por recibir el mayor número de reseñas, con entre 6 y 7 reseñas cada uno, lo que indica un mayor nivel de participación de los lectores. Asimismo, se observa que algunos libros no tienen reseñas de texto, aunque sí cuentan con una calificación promedio, lo que sugiere que varios usuarios prefieren calificar los libros sin dejar comentarios. En general, las calificaciones promedio oscilan entre 3 y 5 estrellas, reflejando una percepción mayoritariamente positiva del catálogo disponible.

## 4. Editorial con más libros de más de 50 páginas

```python
query = """
SELECT
    p.publisher,
    COUNT(*) AS total_books
FROM books AS b
JOIN publishers AS p
ON b.publisher_id = p.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher
ORDER BY total_books DESC
LIMIT 1;
"""

pd.io.sql.read_sql(query, con = engine)
```

El análisis muestra que Penguin Books es la editorial que ha publicado el mayor número de libros con más de 50 páginas, con un total de 42 títulos en la base de datos. Este resultado indica que la editorial tiene una fuerte presencia dentro del catálogo de libros de mayor extensión, excluyendo folletos y publicaciones cortas. Su amplia participación sugiere que Penguin Books es uno de los principales contribuyentes al contenido disponible para los lectores de la plataforma.

## 5. Autor con la calificación promedio más alta (libros con al menos 50 calificaciones)

```python
query = """
SELECT
    a.author,
    ROUND(AVG(r.rating),2) AS average_rating
FROM authors AS a

JOIN books AS b
ON a.author_id = b.author_id

JOIN ratings AS r
ON b.book_id = r.book_id

WHERE b.book_id IN
(
    SELECT
        book_id
    FROM ratings
    GROUP BY book_id
    HAVING COUNT(*) >= 50
)

GROUP BY a.author
ORDER BY average_rating DESC
LIMIT 1;
"""

pd.io.sql.read_sql(query, con = engine)
```

El análisis identificó que J.K. Rowling/Mary GrandPré es el autor con la calificación promedio más alta, alcanzando una puntuación de 4.29 estrellas entre los libros que recibieron al menos 50 calificaciones. Al establecer este umbral mínimo de evaluaciones, el resultado se basa en una muestra representativa de opiniones de los lectores, lo que aumenta la confiabilidad del análisis. Este hallazgo sugiere que las obras de este autor mantienen un alto nivel de satisfacción entre un número considerable de usuarios, consolidándose como una de las colecciones mejor valoradas de la plataforma.

## 6. Promedio de reseñas entre usuarios que calificaron más de 50 libros

```python
query = """
SELECT
    AVG(review_count) AS average_reviews
FROM
(
    SELECT
        username,
        COUNT(*) AS review_count
    FROM reviews
    WHERE username IN
    (
        SELECT
            username
        FROM ratings
        GROUP BY username
        HAVING COUNT(*) > 50
    )
    GROUP BY username
) AS t;
"""

pd.io.sql.read_sql(query, con = engine)
```

El análisis muestra que los usuarios que calificaron más de 50 libros escribieron, en promedio, 24.33 reseñas de texto. Esto indica que, aunque estos usuarios son muy activos al momento de otorgar calificaciones, no registran una reseña escrita para cada libro evaluado. En otras palabras, existe una mayor tendencia a calificar los libros que a dejar comentarios, lo que sugiere que las calificaciones representan una forma de participación más frecuente y menos demandante para los usuarios de la plataforma.
